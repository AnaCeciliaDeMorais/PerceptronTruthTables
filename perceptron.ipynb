```python
import numpy as np
import os
```

Perceptron 1 camada - Tabela AND

```python
entradas = np.array([[0,0],
                     [0,1],
                     [1,0],
                     [1,1]])

saidas = np.array([0,0,
                   0,1])

pesos = np.array([0.0,
                  0.0])
```

Perceptron 1 camada - Tabela OR

```python
entradas1 = np.array([[0,0],
                     [0,1],
                     [1,0],
                     [1,1]])

saidas1 = np.array([0,1,
                   1,1])

pesos1 = np.array([0.0,
                  0.0])
```

Perceptron 1 camada - Tabela XOR

```python
entradas2 = np.array([[0,0],
                     [0,1],
                     [1,0],
                     [1,1]])

saidas2 = np.array([0,1,
                   1,0])

pesos2 = np.array([0.0,
                  0.0])
```

Taxa de aprendizado mantida em 0.5, para evitar erros


```python
taxa_aprendizado = 0.5
```

Funcao soma - multiplica as matrizes e soma os valores em cada celula

```python
def Soma(e,p):
    return e.dot(p)

soma = Soma(entradas, pesos)
soma1= Soma(entradas1, pesos1)
soma2= Soma(entradas2, pesos2)
```

```python
def funcao_degrauAND(soma):
  if (soma >= 1):
    return 1
  return 0
```

```python
def funcao_degrauOR(soma1):
  if (soma1 >= 1):
    return 1
  return 0
```

Multiplica os valores de cada peso pelos contidos nos arrays

```python
def calculo_saidaAND(reg):
  s = reg.dot(pesos)
  return funcao_degrauAND(s)
```

```python
def calculo_saidaOR(reg):
  s1 = reg.dot(pesos1)
  return funcao_degrauOR(s1)
```

```python
def calculo_saidaXOR(reg):
  s2 = reg.dot(pesos2)
  return funcao_degrauXOR(s2)
```

```python
def calcula_e_atualiza():
  erro_total = 1
  while (erro_total != 0):
    erro_total = 0

    for i in range (len(saidas)):
      calc_saidaAND = calculo_saidaAND(np.array(entradas[i]))
      erro = abs(saidas[i] - calc_saidaAND)

      calc_saidaOR = calculo_saidaOR(np.array(entradas1[i]))
      erro1 = abs(saidas1[i] - calc_saidaOR)

      calc_saidaXOR = calculo_saidaXOR(np.array(entradas2[i]))
      erro2 = abs(saidas2[i] - calc_saidaXOR)

      erro_total = erro_total + erro + erro1 + erro2

      for j in range(len(pesos)):
        pesos[j] = pesos[j] + (taxa_aprendizado * entradas[i][j] * erro)
        print('Pesos Atualizados AND > ' + str(pesos[j]))

        pesos1[j] = pesos1[j] + (taxa_aprendizado * entradas1[i][j] * erro1)
        print('Pesos Atualizados OR > ' + str(pesos1[j]))

        pesos2[j] = pesos2[j] + (taxa_aprendizado * entradas2[i][j] * erro2)
        print('Pesos Atualizados XOR > ' + str(pesos2[j]))

    print('Total de Erros: ' + str(erro_total))

calcula_e_atualiza()
```
